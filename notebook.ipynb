## Step 3 — Feature factory (rolling features from FULL history)

**Why full history, not just 2010+?** A driver's *previous 5 races* in early 2010 may include 2009 races. To know those, features must be computed on the whole dataset, then we filter later.

**Key idea — `shift(1)`**: pushes each driver's results down by one race, so the feature at row *R* only sees races strictly **before** *R*. Without it, the current race's result would leak into its own features (the classic leakage bug).

DNF handling: `position` is missing for DNFs, so the rolling mean averages only actual finishes (with `min_periods=1` so debut-ish rows still get a value).

```python
# STEP 3a - build the full-history table
import numpy as np
import pandas as pd

races = pd.read_csv("data/races.csv")
results = pd.read_csv("data/results.csv")
quali = pd.read_csv("data/qualifying.csv")
drivers = pd.read_csv("data/drivers.csv")
constructors = pd.read_csv("data/constructors.csv")

races_small = races[["raceId", "year", "round", "circuitId"]]
full = results.merge(races_small, on="raceId", how="left")
full = full.merge(drivers[["driverId", "driverRef"]].rename(columns={"driverRef": "driver_name"}), on="driverId", how="left")
full = full.merge(constructors[["constructorId", "name"]].rename(columns={"name": "constructor_name"}), on="constructorId", how="left")
full = full.merge(quali[["raceId", "driverId", "position"]].rename(columns={"position": "quali_pos"}), on=["raceId", "driverId"], how="left")

full["podium"] = (full["positionOrder"] <= 3).astype(int)
# results.csv stores missing values as the literal string backslash-N; force them to NaN
full["position"] = pd.to_numeric(full["position"], errors="coerce")
full = full.sort_values(["year", "round"]).reset_index(drop=True)
print("full history:", full.shape)
```

```python
# STEP 3b - rolling features (shift(1) = exclude the current race!)
full["driver_avg_finish_prev5"] = (
    full.groupby("driverId")["position"].shift(1)
        .groupby(full["driverId"]).transform(lambda s: s.rolling(5, min_periods=1).mean()))
full["constructor_avg_finish_prev5"] = (
    full.groupby("constructorId")["position"].shift(1)
        .groupby(full["constructorId"]).transform(lambda s: s.rolling(5, min_periods=1).mean()))
full["driver_podium_rate"] = (
    full.groupby("driverId")["podium"].shift(1)
        .groupby(full["driverId"]).transform(lambda s: s.expanding().mean()))

print(full[["driver_name", "year", "round", "driver_avg_finish_prev5", "driver_podium_rate"]].tail())
```

Notes:
- **Constructor renames** (e.g. Renault -> Alpine got a new `constructorId`) restart team history — acceptable simplification, documented in the README.
- Sanity check: at Hamilton's 2012 round 1, `driver_avg_finish_prev5` = 3.75 which reflects races from 2011 — **not** that race itself.

```python
# STEP 3c - analysis frame: modern era only
df = full[full["year"] >= 2010].copy().reset_index(drop=True)
print("analysis frame:", df.shape)
print("podiums:", df["podium"].sum(), "| rate:", round(df["podium"].mean(), 4))
print("quali_pos NaN:", df["quali_pos"].isna().sum())
```

## Step 4 — Clean: drop leakage, impute missing, encode circuits

- **Leakage list** is explicit — these columns encode the race *outcome* and must never be features: `positionOrder, position, positionText, points, laps, time, milliseconds, fastestLap*, rank, statusId, number`.
- **Impute in meaning order**: `grid == 0` means "no grid time" -> treat as NaN; fill `quali_pos` from grid and vice-versa (they are nearly the same information in modern F1); final fallback = 25 ("backfield placeholder") for true debuts.
- Rolling-feature NaNs (debut races) are filled with the **train-period median** so the test period never influences the fill.
- `circuitId` is one-hot encoded (~30 columns) — safe for trees, kept because circuit effects are real.

```python
# STEP 4 - clean
leak_cols = ["positionOrder", "position", "positionText", "points", "laps", "time", "milliseconds",
             "fastestLap", "fastestLapTime", "fastestLapSpeed", "rank", "statusId", "number"]
df = df.drop(columns=leak_cols, errors="ignore")

df.loc[df["grid"] == 0, "grid"] = np.nan          # grid 0 = "no recorded grid time"
df["quali_pos"] = df["quali_pos"].fillna(df["grid"])
df["grid"] = df["grid"].fillna(df["quali_pos"])
df["quali_pos"] = df["quali_pos"].fillna(25)
df["grid"] = df["grid"].fillna(25)

era = df["year"] <= 2021                          # train-period only stats
df["driver_avg_finish_prev5"] = df["driver_avg_finish_prev5"].fillna(df.loc[era, "driver_avg_finish_prev5"].median())
df["constructor_avg_finish_prev5"] = df["constructor_avg_finish_prev5"].fillna(df.loc[era, "constructor_avg_finish_prev5"].median())
df["driver_podium_rate"] = df["driver_podium_rate"].fillna(df.loc[era, "driver_podium_rate"].median())

print("NaNs left in numeric features:",
      df[["grid", "quali_pos", "driver_avg_finish_prev5",
          "constructor_avg_finish_prev5", "driver_podium_rate"]].isna().sum().sum())

df = pd.concat([df, pd.get_dummies(df["circuitId"].astype(int), prefix="circuit").astype(int)], axis=1)
print("after one-hot circuits:", df.shape)
```

## Step 5 — Time-based train/test split

Train = 2010-2021, Test = 2022-2024. **No random split**: a real forecasting question is "predict the future from the past", so the test set must live strictly *after* the training period. A random split would let 2023 rows teach the model about 2022 rows.

```python
# STEP 5 - split by year
num_features = ["grid", "quali_pos", "driver_avg_finish_prev5",
                "constructor_avg_finish_prev5", "driver_podium_rate"]
circ_cols = [c for c in df.columns if c.startswith("circuit_")]
feature_cols = num_features + circ_cols

train = df[df["year"] <= 2021]
test  = df[df["year"] >= 2022]
X_train, y_train = train[feature_cols], train["podium"]
X_test,  y_test  = test[feature_cols],  test["podium"]
print("TRAIN", X_train.shape, "| podium rate:", round(y_train.mean(), 4))
print("TEST ", X_test.shape,  "| podium rate:", round(y_test.mean(), 4))
```

## Step 6 — Baseline: "podium if grid <= 3"

The dumbest sensible rule. Any real model must beat it or it is decoration. Note its strengths (P1-P3 starters are genuinely likely to podium) and its blindness (it never predicts a hero-drive from P7, and thinks pole-sitters never crash).

```python
# STEP 6 - baseline
from sklearn.metrics import (precision_score, recall_score, f1_score, accuracy_score,
                             roc_auc_score, confusion_matrix, classification_report)

base_pred = (test["grid"] <= 3).astype(int).values
print(classification_report(y_test, base_pred, digits=3))
print(confusion_matrix(y_test, base_pred))
```

## Step 7 — Models: Logistic Regression vs Random Forest

- **LogReg** gets `StandardScaler` inside a Pipeline (fit on *train only* — never the test years) and `class_weight="balanced"` because podiums are the rare class (~14%).
- **Random Forest** needs no scaling and catches non-linear patterns.
- Metrics: precision ("when we say podium, how often right?"), recall ("of all real podiums, how many caught?"), F1 (their balance), ROC-AUC (ranking quality across all thresholds). Accuracy is reported but *"always say no"* would score ~85%, which is exactly why it is not the headline.

```python
# STEP 7 - models + metrics table
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

lr = make_pipeline(StandardScaler(), LogisticRegression(max_iter=2000, class_weight="balanced"))
lr.fit(X_train, y_train)

rf = RandomForestClassifier(n_estimators=400, class_weight="balanced", random_state=42, n_jobs=-1)
rf.fit(X_train, y_train)

def metrics_row(name, y_true, y_pred, y_proba=None):
    return {"model": name,
            "precision": round(precision_score(y_true, y_pred), 3),
            "recall": round(recall_score(y_true, y_pred), 3),
            "F1": round(f1_score(y_true, y_pred), 3),
            "accuracy": round(accuracy_score(y_true, y_pred), 3),
            "roc_auc": round(roc_auc_score(y_true, y_proba), 3) if y_proba is not None else None}

metrics_df = pd.DataFrame([
    metrics_row("baseline grid<=3", y_test, base_pred),
    metrics_row("logistic regression", y_test, lr.predict(X_test), lr.predict_proba(X_test)[:, 1]),
    metrics_row("random forest", y_test, rf.predict(X_test), rf.predict_proba(X_test)[:, 1]),
]).set_index("model")
metrics_df
```

```python
# STEP 7b - what did the models learn?
# Logistic Regression: one-unit effect of each scaled feature on log-odds
coefs = pd.Series(lr[-1].coef_[0], index=feature_cols)
print("LogReg numeric effects (scaled):")
print(coefs[num_features].sort_values(ascending=False).round(3).to_string())
print()
imp = pd.Series(rf.feature_importances_, index=feature_cols).sort_values(ascending=False)
print("RF top-8 importances:")
print(imp.head(8).round(3).to_string())
```

## Step 8 — Overfitting demo: decision tree depth 1 to 20

Small depth = model too simple (both scores low). Huge depth = model memorises the training noise: train accuracy climbs to 1.0 while test accuracy plateaus/drops. The widening **gap** between the two curves *is* overfitting made visible.

```python
# STEP 8 - train vs test score as depth grows
from sklearn.tree import DecisionTreeClassifier
import matplotlib.pyplot as plt

tr_acc, te_acc = [], []
for d in range(1, 21):
    t = DecisionTreeClassifier(max_depth=d, random_state=42).fit(X_train, y_train)
    tr_acc.append(t.score(X_train, y_train))
    te_acc.append(t.score(X_test, y_test))

plt.figure(figsize=(8, 5))
plt.plot(range(1, 21), tr_acc, label="train accuracy")
plt.plot(range(1, 21), te_acc, label="test accuracy")
plt.xlabel("max_depth")
plt.ylabel("accuracy")
plt.title("Overfitting demo: growing tree depth")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
print("depth 5: train", round(tr_acc[4], 3), "test", round(te_acc[4], 3))
print("depth 20: train", round(tr_acc[19], 3), "test", round(te_acc[19], 3))
```

## Step 9 — Unsupervised: PCA + k-means on driver careers

One row per driver (races >= 20): career stats -> StandardScaler -> PCA to 2D -> k-means with k chosen by silhouette. Then we *name* the clusters by reading their centres and famous members.

```python
# STEP 9 - unsupervised learning
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

ds = full.groupby(["driverId", "driver_name"]).agg(
    races=("resultId", "count"), podiums=("podium", "sum"),
    avg_finish=("position", "mean")).reset_index()
ds = ds[ds["races"] >= 20].copy()                  # filter one-off drivers
ds["avg_finish"] = ds["avg_finish"].fillna(ds["avg_finish"].median())
ds["podium_rate"] = ds["podiums"] / ds["races"]

Xs = StandardScaler().fit_transform(ds[["races", "podium_rate", "avg_finish"]])
pca = PCA(n_components=2).fit(Xs)
print("PCA explained variance:", pca.explained_variance_ratio_.round(3),
      "| total:", round(pca.explained_variance_ratio_.sum(), 3))

sils = {k: silhouette_score(Xs, KMeans(n_clusters=k, n_init=10, random_state=42).fit_predict(Xs))
        for k in range(2, 9)}
print("silhouette by k:", {k: round(v, 3) for k, v in sils.items()})
best_k = max(sils, key=sils.get)
ds["cluster"] = KMeans(n_clusters=best_k, n_init=10, random_state=42).fit_predict(Xs)
print("best k =", best_k)
summary = ds.groupby("cluster")[["races", "podium_rate", "avg_finish"]].mean().round(2)
print(summary.to_string())
print()
top6 = (ds.sort_values("races", ascending=False).groupby("cluster").head(6)
        .sort_values(["cluster", "races"], ascending=[True, False]))
print(top6[["cluster", "driver_name", "races", "podium_rate", "avg_finish"]].round(2).to_string(index=False))
ds.to_csv("driver_clusters.csv", index=False)
```

```python
# STEP 9b - see the 2D picture
plt.figure(figsize=(8, 6))
plt.scatter(ds["cluster"], ds["podium_rate"], s=ds["races"] / 4, alpha=0.5)
plt.xlabel("cluster"); plt.ylabel("podium rate"); plt.title("Driver clusters (step 9)")
plt.grid(alpha=0.3)
plt.show()
# Reminder: PCA is 2D, so a nicer scatter (optional) would use pca coordinates.
```
